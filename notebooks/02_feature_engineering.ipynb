```python
import pandas as pd
import numpy as np
from pathlib import Path

PROCESSED_DIR = Path("../data/processed")

regions = [
    "AEP", "COMED", "DAYTON", "DEOK", "DOM",
    "DUQ", "EKPC", "FE", "NI", "PJME", "PJMW"
]

data = {}

for region in regions:
    path = PROCESSED_DIR / f"{region}_cleaned.csv"
    df = pd.read_csv(path)
    df["Datetime"] = pd.to_datetime(df["Datetime"])
    data[region] = df

print(f"Loaded {len(data)} regions")
```

```python
feature_data = {}

for region, df in data.items():
    df = df.copy()

    df["hour"] = df["Datetime"].dt.hour
    df["day"] = df["Datetime"].dt.day
    df["day_of_week"] = df["Datetime"].dt.dayofweek
    df["month"] = df["Datetime"].dt.month
    df["year"] = df["Datetime"].dt.year

    df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

    feature_data[region] = df

print("Time features created.")
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"


for region, df in feature_data.items():
    df["season"] = df["month"].apply(get_season)
```

```python
for region, df in feature_data.items():

    energy_col = df.columns[1]

    df["lag_1"] = df[energy_col].shift(1)
    df["lag_2"] = df[energy_col].shift(2)
    df["lag_3"] = df[energy_col].shift(3)

    df["lag_24"] = df[energy_col].shift(24)
    df["lag_48"] = df[energy_col].shift(48)

    df["lag_168"] = df[energy_col].shift(168)
```

```python
for region, df in feature_data.items():

    energy_col = df.columns[1]

    df["rolling_mean_24"] = (
        df[energy_col]
        .shift(1)
        .rolling(24)
        .mean()
    )

    df["rolling_std_24"] = (
        df[energy_col]
        .shift(1)
        .rolling(24)
        .std()
    )

    df["rolling_mean_168"] = (
        df[energy_col]
        .shift(1)
        .rolling(168)
        .mean()
    )
```
